```python
def run_noi_pipeline(
    my_client,
    noi,
    leftRight,
    Cnumber,
    prePost,
    nFilterStr,
    nSys,
    neuronsToExclude=None
    ):
    """
    Runs the full NLP query pipeline for a given neuron of interest (NOI).
    """

    # Build strings
    nFilter = f"/r(.*){nFilterStr}(.*)(.*)/r"
    noiFileName = f"{noi}_{leftRight}_{nFilterStr}_{nSys}{prePost}_001"
    print(noiFileName)
    NOIqueryString = f'add /r(.*){noi}(.*){leftRight}(.*){Cnumber}(.*)/r'
    print(NOIqueryString)
    PrePostSynQueryString = f"add {prePost}synaptic {nFilter} neurons with at least {nSys} synapses"
    print(PrePostSynQueryString)
    


    # --- Execute in order (each call blocks until complete) ---
    results = {}

    results["clear"] = my_client.executeNLPquery("clear")
    results["NOI"] = my_client.executeNLPquery(NOIqueryString)
    results["prepost"] = my_client.executeNLPquery(PrePostSynQueryString)

    if neuronsToExclude:
        ExcludeQueryString = f'remove /r(.*){neuronsToExclude}(.*)/r'
        print(ExcludeQueryString)
        results["exclude"] = my_client.executeNLPquery(ExcludeQueryString)
        

    return noiFileName, results, nFilter
```

# Begin Synaptic data aggregation

```python


# ⏱️ Timer start
start_time = time.time()

neurons = my_client.NLP_result.neurons
hashids = list(neurons.keys())
names = [neurons[k]['uname'] for k in hashids]

# 🧠 Synapse collection
preSynapses = []
postSynapses = []

for i, hashid in enumerate(hashids):
    res = my_client.getInfo(hashid)
    data = res["data"]

    if "connectivity" in data:
        presyn = data["connectivity"]["pre"]["details"]
        postsyn = data["connectivity"]["post"]["details"]

        for syn in presyn:
            preSynapses.append(("pre", syn["syn_uname"], names[i], syn["uname"],
                                syn["s_rid"], syn["number"], True))

        for syn in postsyn:
            postSynapses.append(("post", syn["syn_uname"], names[i], syn["uname"],
                                 syn["s_rid"], syn["number"], True))

# 🧩 NumPy structured arrays
dtype = [('type', 'U4'), ('syn_uname', 'U100'), ('neuron_name', 'U100'),
         ('uname', 'U100'), ('s_rid', 'U100'), ('number', 'f8'), ('valid', 'bool')]

pre_array = np.array(preSynapses, dtype=dtype)
post_array = np.array(postSynapses, dtype=dtype)

# 🔍 Filter out 'segment' entries efficiently
pre_array = pre_array[np.char.find(pre_array['uname'], 'segment') == -1]
post_array = post_array[np.char.find(post_array['uname'], 'segment') == -1]

connectivity = {"hashids": hashids, "names": names, "pre_array": pre_array, "post_array": post_array}

# 📊 Dictionary construction
postSynNumberDict = {}
preSynNumberDict = {}

def process_array(array, synNumberDict, synType):
    neuron_names = np.unique(array['neuron_name'])
    for neuron in neuron_names:
        mask = array['neuron_name'] == neuron
        syn_data = array[mask]
        syn_numbers = syn_data['number']

        if syn_numbers.size == 0:
            continue

        z_scores = stats.zscore(syn_numbers)

        if np.any(np.isnan(z_scores)):
            array['valid'][mask] = False
            print(f'\033[93m{neuron} Error: Nan Value Found in array\033[0m')
            continue

        zmax = np.max(z_scores)
        norm_scores = 1 - ((zmax - z_scores) / zmax)

        print(f'\033[92m{neuron}: {synType}SynZmax= {zmax}\033[0m')

        for uname, rid, z, norm in zip(syn_data['syn_uname'], syn_data['s_rid'], z_scores, norm_scores):
            synNumberDict[rid] = [uname, z, norm]

process_array(connectivity["post_array"], postSynNumberDict, "post")
process_array(connectivity["pre_array"], preSynNumberDict, "pre")

# 🕒 Timing metrics
numberOfNodes = len(names)
end_time = time.time()
elapsed_time = end_time - start_time

avSynCalc_time = elapsed_time / (len(pre_array) + len(post_array))
avNeuronCalc_time = elapsed_time / numberOfNodes

avSynCalc_time = 10

if elapsed_time > avSynCalc_time:
    gvLayoutEngine = 'neato'
    gGraphvizMode = 'sgd'
else:
    gvLayoutEngine = 'sfdp'
    gGraphvizMode = 'spring'

print ("No. of HashIDs= " + str( len(hashids)))
print ("No. of names= " + str( numberOfNodes ))
print ("No. of preSynapses= " + str( len(preSynapses)))
print ("No. of postSynapses= " + str( len(postSynapses)))
print(f"Elapsed time: {elapsed_time:.2f} seconds")
print(f"Avg. time per synapse: {avSynCalc_time:.4f} seconds")
print(f"Avg. time per neuron: {avNeuronCalc_time:.4f} seconds")
print(f"Graphviz  set to Engine: {gvLayoutEngine} and Mode:{gGraphvizMode} ")
```

```python
def add_graph_edge(g, head, tail, zscore=None,
                    penwidth=None,
                    color=None,
                    edgeFontcolor="#40e0d0",
                    edgeFontName = "Ariel",
                    edgeFontSize = "5.0",
                    labelfloat = "false",
                    splines = "true",
                    arrowsize="0.4",
                    weight="3",
                    label_prefix="Zscore = ",
                    zScoreMultiplyer=2,
                    calcColorFn=None):
    """
    Adds a styled edge to the Graphviz graph object `g`.

    Parameters:
        g (Digraph): your graph object
        head (str): source node name
        tail (str): destination node name
        zscore (float): value to encode visually (used to compute penwidth and color)
        penwidth (float or str): overrides computed penwidth if provided
        color (str): overrides computed color if provided
        fontcolor (str): label font color
        fontname (str): label font name
        fontsize (str): label font size
        arrowsize (str): edge arrow size
        weight (str): edge weight
        label_prefix (str): prefix for edge label text
        labelfloat (str): whether label floats with edge curvature
        zScoreMultiplyer (float): multiplier for visual scaling
        calcColorFn (function): function to compute [penwidth, color] from zscore
    """

    # Auto-compute visual attributes if not provided
    if zscore is not None and calcColorFn:
        computed = calcColorFn(zscore)
        pw = penwidth if penwidth is not None else computed[0] * zScoreMultiplyer
        pc = color if color is not None else computed[1]
        label = label_prefix + str(round(computed[0], 2))
    else:
        pw = penwidth or "1.0"
        pc = color or "gray"
        label = ""
    #print(f"\033[92mcreating edge with head:\033[38;5;208m{head}\033[92m -> \033[96m{tail}\033[0m")
    g.edge(str(head), str(tail),
           labelfloat=labelfloat,
           label=label,
           fontsize=edgeFontSize,
           fontcolor=edgeFontcolor,
           fontname=edgeFontName,
           penwidth=str(pw),
           weight=weight,
           arrowsize=arrowsize,
           color=pc)
    
def add_edge_tograph(graph, head_name, tail_name, zscore, color_fn):
    """
    Wrapper for add_graph_edge to simplify synaptic connection additions.

    Parameters:
        graph: Graph object to modify.
        pre_name: Node name representing presynaptic neuron.
        post_name: Node name representing postsynaptic neuron.
        zscore: Z-score value representing synaptic weight.
        color_fn: Function to compute edge color from z-score.

    Returns:
        None
    """
    add_graph_edge(
        g=graph,
        head=head_name,
        tail=tail_name,
        zscore=zscore,
        calcColorFn=color_fn
    )
```

```python
import warnings

def clean_node_id(node_str):
    return node_str.strip('"').strip()

def parse_edge_line(line):
    pattern = r'\"?([^\s\"]+)\"?\s*->\s*\"?([^\s\"]+)\"?.*?label=\"Zscore\s*=\s*([-+]?[0-9]*\.?[0-9]+)\"'
    match = re.search(pattern, line.strip())
    if match:
        head, tail, z_score = match.groups()
        return head.strip('"'), tail.strip('"'), float(z_score), line.strip()
    return None


def merge_graph_files(directory='merge_graphs', output_file='merged_data.gv'):
    if not os.path.exists(directory):
        os.makedirs(directory)
        warnings.warn("Directory 'merge_graphs' not found. Created empty directory. Halting routine.")
        return

    merged_edges = []
    graph_header = []

    for filename in os.listdir(directory):
        if filename.endswith('.gv') and not filename.startswith('merged') and not filename.startswith('unused'):
            with open(os.path.join(directory, filename), 'r') as f:
                lines = f.readlines()

            if not graph_header:
                graph_header = lines[:2]

            body_lines = []  # 🔄 Reset for each file
            layout_directives = ('size=', 'rotation=', 'splines=')

            for line in lines[2:]:
                stripped = line.strip()
                if stripped == '}':
                    continue
                # Only skip lines that *start* with layout directives
                if any(stripped.startswith(directive) for directive in layout_directives):
                    print(f"Skipping layout directive line: {stripped}")
                    continue
                body_lines.append(line)

                
            merged_edges.extend(body_lines)
            print(f"{len(body_lines)} lines of file {filename} merged. Total lines in merged now equals: {len(merged_edges)}")

    # Deduplicate after all files are processed
    edge_dict = {}
    unused_edges = {}
    node_buffer = []
    node_set = set()

    for line in merged_edges:
        if line.lstrip().startswith('#'):
            continue  # Skip discarded-edge log lines

        parsed = parse_edge_line(line)

        if parsed:
            #print(f"Parsed result: --{parsed}--, Line = {line}")
            parsed_head, parsed_tail, z, full_line = parsed
            key = (parsed_head, parsed_tail)

            # Filtering logic: keep only highest Z-score
            if key not in edge_dict or z > edge_dict[key][0]:
                edge_dict[key] = (z, full_line)
            else:
                # Log discarded edge
                unused_edges.setdefault(key, []).append((z, full_line))
        else:
            # Node handling logic (as discussed earlier)
            line = line.strip()
            if not line:
                continue
            node_buffer.append(line)
            if line.endswith(']'):
                full_node = ' '.join(node_buffer)
                node_set.add(full_node)
                node_buffer = []
                #print(f"Parsed Node found: {full_node}")



                
    output_path_merged = os.path.join(directory, output_file)

    #pdb.set_trace()
    #Write out data to merged file
    with open(output_path_merged, 'w') as out:
        out.writelines(graph_header)
        # Write unique nodes
        for node_line in sorted(node_set):
            out.write(node_line + '\n')
        # Write edges
        for _, (_, line) in edge_dict.items():
            out.write(line + '\n')
        out.write('\tsize="50,50"\n')
        out.write('}\n')

    print(f"✅ Merged graph written to {output_file} with {len(edge_dict)} unique edges.")
    unsuedLogFile = f"{noiFileName}_unused_log.gv"

    output_path_log = os.path.join(directory, unsuedLogFile)
    
    with open(output_path_log, 'w') as f:
        for key, entries in unused_edges.items():
            for z_val, line_text in entries:
                f.write(f"# Discarded [{key[0]} -> {key[1]}] with z={z_val}: {line_text}\n")
    print(f"✅ Unused itmes log written to {output_path_log} with {len(unused_edges)}  items.")

############RENAME IF Needed
# currentMergedFileName = 'merged_data.gv'
currentMergedFileName = f'{noiFileName}_merged_data_v1.gv'

merge_graph_files(output_file=currentMergedFileName)
```

```python
def extract_zscore(line):
    match = re.search(r'label="[^=]+=\s*([0-9.]+)"', line)
    return float(match.group(1)) if match else 0.0

def find_synapse_legs(synapse, edge_lines):
    """Find both legs of a synapse: pre and post edges."""
    legs = []
    for line in edge_lines:
        parts = line.strip().split('->')
        if len(parts) != 2:
            raise ValueError(f"Malformed edge line at index {i}: {line.strip()}")
            continue
        src = parts[0].strip().strip('"')
        tgt = parts[1].split()[0].strip().strip('"')
        if src == synapse or tgt == synapse:
            z = extract_zscore(line)
            legs.append((src, tgt, z, line))
    return legs

def get_base_graph_file(directory='merge_graphs', override=None):
    """Return the path to the base graph file, either via override or latest merged_*.gv."""
    if override:
        path = os.path.join(directory, override)
        if not os.path.exists(path):
            raise FileNotFoundError(f"Override file '{override}' not found in {directory}")
        return path

    merged_files = [f for f in os.listdir(directory) if 'merged' in f and f.endswith('.gv')]
    if not merged_files:
        raise FileNotFoundError("No merged_*.gv files found in the directory.")
    
    merged_files.sort(key=lambda f: os.path.getmtime(os.path.join(directory, f)), reverse=True)
    return os.path.join(directory, merged_files[0])


def filter_graph(directory='merge_graphs', threshold=0.5):
    # Load lines
    with open(get_base_graph_file(directory), 'r') as f:
        lines = f.readlines()

    # Separate node and edge lines
    node_lines = [line for line in lines if '->' 
              not in line 
              and '[' in line and not line.strip().startswith(('graph [', 'node [', 'edge ['))]
    
    edge_lines = [line for line in lines if '->' in line]

    # Extract all node IDs and synapse nodes
    all_nodes = set()
    synapse_nodes = set()
    node_dict = {}

    for line in node_lines:
        tokens = line.strip().split()
        if tokens:
            node_id = tokens[0].strip('"')
            all_nodes.add(node_id)
            node_dict[node_id] = line
            if '--' in node_id:
                synapse_nodes.add(node_id)

    accepted_edges = []
    nodes_in_accepted_edges = set()
    rejected_synapses = set()

    # Evaluate each synapse as a unit
    for synapse in synapse_nodes:
        legs = find_synapse_legs(synapse, edge_lines)

        if len(legs) != 2:
            print(f"⚠️ Synapse {synapse} has {len(legs)} legs, skipping.")
            continue

        z_scores = [leg[2] for leg in legs]
        if any(z >= threshold for z in z_scores):
            for src, tgt, z, line in legs:
                accepted_edges.append(line)
                nodes_in_accepted_edges.update([src, tgt])
        else:
            rejected_synapses.add(synapse)

    # Filter nodes based on accepted edges
    filtered_nodes = [
        line for node, line in node_dict.items()
        if node in nodes_in_accepted_edges and node not in rejected_synapses
    ]

    # Optional: log dropped nodes
    dropped_nodes = all_nodes - nodes_in_accepted_edges - rejected_synapses
    if dropped_nodes:
        print(f"Dropped {len(dropped_nodes)} nodes: {sorted(dropped_nodes)}")

    return filtered_nodes, accepted_edges

if __name__ == "__main__":
    directory = 'merge_graphs'
    override_filename = None  # e.g. 'merged_subset.gv' if you want to override
    output_file = 'filtered_output.gv'
    
    #######################################
    ######### Set Threshold ###############
    #######################################
    
    tempThreshold = 0.3

    try:
        filtered_nodes, accepted_edges = filter_graph(directory, threshold=tempThreshold)
        print(f"✅ Filtered graph written to {output_file} with {len(filtered_nodes)} nodes and {len(accepted_edges)} edges.")
        
    except Exception as e:
        print(f"❌ Error: {e}")
        traceback.print_exc()

        

    grouped_nodes = defaultdict(list)
    ungrouped_nodes = []

    for line in filtered_nodes:
        tokens = line.strip().split()
        if not tokens:
            continue

        node_id = tokens[0].strip('"')

        if '--' in node_id:
            # ✅ Synapse node: keep for output, but don't group
            ungrouped_nodes.append(line)
        else:
            # ✅ Neuron node: group by prefix
            prefix = node_id[:3]
            grouped_nodes[prefix].append(line)

        
        
    
    # Write output file with header and footer
    with open(input_file, 'r') as f:
        lines = f.readlines()
    graph_header = lines[:2]
    graph_footer = lines[-2:]

    # Write output
    with open(output_file, 'w') as out:
        out.writelines(graph_header)

        # ✅ All writes must be inside this block
        for prefix in sorted(grouped_nodes):
            out.write(f"subgraph cluster_{prefix} {{\n")
            out.write(f'    label = "{prefix} Group";\n')
            out.write("    style = none;\n    color = black;\n ")
            for node_line in sorted(grouped_nodes[prefix]):
                out.write("    " + node_line.strip() + '\n')
            out.write("}\n")
            
        # ✅ Write synapse nodes outside any group
        for node_line in sorted(ungrouped_nodes):
            out.write("    " + node_line.strip() + '\n')

        for edge_line in sorted(set(accepted_edges)):
            out.write(edge_line.strip() + '\n')

        out.writelines(graph_footer)

```

```python
#In this step we gather all the Neurotransmitter data from flywire
#This can be only done for each type 

def load_neurotransmitter_dict(filename="neuronNTdict.json"):
    """
    Loads the neurotransmitter dictionary from a JSON file if it exists.

    Returns:
    - dict: The loaded neurotransmitter dictionary, or empty dict if not found.
    """
    if os.path.exists(filename):
        try:
            with open(filename, 'r', encoding='utf-8') as f:
                data = json.load(f)
            print(f"{GREEN}[INFO] Loaded existing neurotransmitter dictionary from '{filename}'{RESET}")
            return data
        except Exception as e:
            print(f"{RED}[ERROR] Failed to load JSON file '{filename}'{RESET}")
            print(f"{RED}        Exception type: {type(e).__name__}{RESET}")
            print(f"{RED}        Exception message: {str(e)}{RESET}")
    else:
        print(f"{YELLOW}[WARN] No existing neurotransmitter file found — starting fresh{RESET}")
    return {}



def save_neurotransmitter_dict(data_dict, filename="neuronNTdict.json", overwrite=True):
    if not overwrite and os.path.exists(filename):
        print(f"{YELLOW}[WARN] File '{filename}' already exists. Skipping save.{RESET}")
        return

    try:
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(data_dict, f, indent=4, ensure_ascii=False)
        print(f"{GREEN}[INFO] Neurotransmitter dictionary saved to '{filename}'{RESET}")
    except Exception as e:
        print(f"{RED}[ERROR] Failed to save JSON file '{filename}'{RESET}")
        print(f"{RED}        Exception type: {type(e).__name__}{RESET}")
        print(f"{RED}        Exception message: {str(e)}{RESET}")
        
def update_neurotransmitter_entry(currNeuron, query_result, neuroTransmitterDict, tauDict, defaultNTdict, GREEN, RED, YELLOW, RESET):
    """
    Updates the neurotransmitter dictionary for a given neuron using Flywire query results.

    Parameters:
    - currNeuron (str): Name of the neuron to update.
    - query_result: Result object from Flywire SQL query.
    - neuroTransmitterDict (dict): Dictionary to update.
    - tauDict (dict): Mapping of neurotransmitters to tau values.
    - defaultNTdict (dict): Fallback NT dictionary if query fails.
    - GREEN, RED, YELLOW, RESET (str): ANSI color codes for print formatting.
    """
    transmitterDict = {}

    try:
        flywireData = [n.out('HasData') for n in query_result.node_objs]
        transmitterList = flywireData[0][1].props['Transmitters']
        for transmitter in transmitterList:
            transmitterDict[transmitter] = tauDict[transmitter]
        neuroTransmitterDict[currNeuron] = transmitterDict
        for transmitter in transmitterList:
            print(f"{GREEN}[INFO] Added {currNeuron} transmitter '{transmitter}' with tau {tauDict[transmitter]}{RESET}")
    except (IndexError, KeyError) as e:
        print(f"{RED}[ERROR] Failed to extract 'Transmitters' for neuron '{currNeuron}' from query result{RESET}")
        print(f"{RED}        Exception type: {type(e).__name__}{RESET}")
        print(f"{RED}        Exception message: {str(e)}{RESET}")
        neuroTransmitterDict[currNeuron] = defaultNTdict
        print(f"{YELLOW}[WARN] Fallback to default neurotransmitter for '{currNeuron}'{RESET}")

def query_and_update_neuron_by_field(field, label, neuroTransmitterDict, tauDict, defaultNTdict, GREEN, RED, YELLOW, RESET):
    """
    Queries Flywire for a neuron using a specified field ('name' or 'uname') and updates the NT dictionary.

    Parameters:
    - field (str): The field to match in the query ('name' or 'uname').
    - label (str): The value to match.
    - neuroTransmitterDict (dict): Dictionary to update.
    - tauDict (dict): Mapping of NTs to tau values.
    - defaultNTdict (dict): Fallback NT dictionary.
    - GREEN, RED, YELLOW, RESET (str): ANSI color codes.
    """

    if label not in neuroTransmitterDict:
        try:
            safe_label = label.replace("'", "''")
            query = f"select from Neuron where {field} matches '{safe_label}'"
            query_result = flywire.sql_query(query)

            if len(query_result) == 0:
                print(f"{YELLOW}[WARN] No results for query '{query}' — assuming acetylcholine{RESET}")
                neuroTransmitterDict[label] = defaultNTdict
            else:
                update_neurotransmitter_entry(label,
                                              query_result,
                                              neuroTransmitterDict,
                                              tauDict, defaultNTdict,
                                              GREEN, RED, YELLOW, RESET)
        except Exception as e:
            print(f"{RED}[ERROR] Query failed for label '{label}' using field '{field}'{RESET}")
            print(f"{RED}        {type(e).__name__}: {str(e).splitlines()[0]}{RESET}")
            neuroTransmitterDict[label] = defaultNTdict
            print(f"{YELLOW}[WARN] Fallback to default neurotransmitter for '{label}' due to query error{RESET}")





defaultNTdict = {'TBD': 2}
neuroTransmitterDict = load_neurotransmitter_dict()


for neuronNode in nodes:
    transmitterDict = {}
    if nodes[neuronNode]["class"] == 'Neuron':
        name = nodes[neuronNode]["name"]
        uname = nodes[neuronNode]["uname"]

        query_and_update_neuron_by_field("name", name, neuroTransmitterDict, tauDict, defaultNTdict, GREEN, RED, YELLOW, RESET)
        query_and_update_neuron_by_field("uname", uname, neuroTransmitterDict, tauDict, defaultNTdict, GREEN, RED, YELLOW, RESET)

                

save_neurotransmitter_dict(neuroTransmitterDict)
display(JSON(neuroTransmitterDict))

```

```python


def replace_tbd_in_file(filepath, name_substring, old_transmitter, new_transmitter, new_tau):
    """
    Loads a neurotransmitter dictionary from file, replaces orignal NT entries for matching neurons,
    backs up the original file, and saves the updated dictionary.

    Parameters:
    - filepath (str): Path to the JSON file.
    - name_substring (str): Substring to match in neuron names.
    - new_transmitter (str): Transmitter to replace 'TBD' with.
    - new_tau (float or int): Tau value to associate with the new transmitter.
    """
    # Load original data
    with open(filepath, 'r') as f:
        neuroTransmitterDict = json.load(f)

    updated_count = 0
    for neuron, nt_dict in neuroTransmitterDict.items():
        if old_transmitter in nt_dict and name_substring in neuron:
            old_tau = nt_dict[old_transmitter]
            del nt_dict[old_transmitter]
            nt_dict[new_transmitter] = new_tau
            updated_count += 1
            print(f"✅ Updated '{neuron}': {old_transmitter} → {new_transmitter} (tau={new_tau}) [was tau={old_tau}]")

    # Backup original file
    backup_path = filepath.replace(".json", "_old.json")
    shutil.copy(filepath, backup_path)
    print(f"\n📁 Original file backed up as: {backup_path}")

    # Save updated data
    with open(filepath, 'w') as f:
        json.dump(neuroTransmitterDict, f, indent=2)
    print(f"💾 Updated dictionary saved to: {filepath}")
    print(f"🔧 Total neurons updated: {updated_count}")
```

```python
def get_nt_keys_by_partial_name(partial_name, nt_dict):
    """
    Returns a list of neurotransmitter keys for any neuron whose name contains the partial_name.
    """
    for full_name, nt_info in nt_dict.items():
        if partial_name in full_name:
            return list(nt_info.keys())
    return []



def parse_dot_file(dot_filename):
    """
    Parses a DOT file and returns a directed NetworkX graph with Z-score annotations.
    """
    graphs = pydot.graph_from_dot_file(dot_filename)
    pydot_graph = graphs[0]
    nx_graph = nx.DiGraph(nx.nx_pydot.from_pydot(pydot_graph))

    # Extract Z-score from node labels
    for node in nx_graph.nodes:
        label = nx_graph.nodes[node].get("label", "")
        match = re.search(r"cent=\s*([0-9.]+)", label)
        if match:
            nx_graph.nodes[node]["zscore"] = float(match.group(1))
        else:
            nx_graph.nodes[node]["zscore"] = None  # fallback

    return nx_graph



import json
import re
import networkx as nx
import pydot

def find_synaptic_motifs(dot_filename, ntDictFile="neuronNTdict.json", outputFile=None):
    with open(ntDictFile, 'r') as f:
        neuroTransmitterDict = json.load(f)

    motifDict = {}
    G = parse_dot_file(dot_filename)
    seen_motifs = set()

    for synapse in G.nodes:
        if '--' in synapse:
            try:
                a, b = synapse.split('--')
            except ValueError:
                continue  # malformed synapse name

            reciprocal = f"{b}--{a}"
            canonical_motif = f"{min(a, b)}--{max(a, b)}"
            if reciprocal in G.nodes and canonical_motif not in seen_motifs:
                nt_a = get_nt_keys_by_partial_name(a, neuroTransmitterDict)
                nt_b = get_nt_keys_by_partial_name(b, neuroTransmitterDict)
                for nt1 in nt_a:
                    for nt2 in nt_b:
                        if nt1 in NTexcInhDict and nt2 in NTexcInhDict:
                            if NTexcInhDict[nt1] != NTexcInhDict[nt2]:
                                motifDict[canonical_motif] = {
                                    "motifType": 'wilson-cowan',
                                    "neuron_A": a,
                                    "neuron_B": b,
                                    "NT_A": nt1,
                                    "NT_B": nt2,
                                    "Z_A": G.nodes[a].get("zscore"),
                                    "Z_B": G.nodes[b].get("zscore"),
                                    "synapse": synapse,
                                    "reciprocal_synapse": reciprocal
                                }
                                seen_motifs.add(canonical_motif)

    print(f"{GREEN}[INFO] Found {len(motifDict)} motifs with opposing NT polarity and reciprocal synapses{RESET}")

    if outputFile:
        with open(outputFile, 'w') as f_out:
            json.dump(motifDict, f_out, indent=2)
        print(f"{GREEN}[INFO] Motif dictionary saved to {outputFile}{RESET}")

    return motifDict


# def find_synaptic_motifs(dot_filename, ntDictFile="neuronNTdict.json"):

#     with open(ntDictFile, 'r') as f:
#         neuroTransmitterDict = json.load(f)
    
    
#     motifDict = {}
#     G = parse_dot_file(dot_filename)

#     for synapse in G.nodes:
#         if '--' in synapse:
#             try:
#                 a, b = synapse.split('--')
#                 #if "APL" in a and "PPL" in b: pdb.set_trace()
#             except ValueError:
#                 continue  # malformed synapse name

#             reciprocal = f"{b}--{a}"
#             if reciprocal in G.nodes:
#                 nt_a = get_nt_keys_by_partial_name(a, neuroTransmitterDict)
#                 nt_b = get_nt_keys_by_partial_name(b, neuroTransmitterDict)
#                 for nt1 in nt_a:
#                     for nt2 in nt_b:
#                         if nt1 in NTexcInhDict and nt2 in NTexcInhDict:
#                             if NTexcInhDict[nt1] != NTexcInhDict[nt2]:
#                                 canonical_motif = f"{min(a, b)}--{max(a, b)}"
#                                 if canonical_motif not in motifDict:
#                                     motifDict[canonical_motif] = {
#                                     "motifType": 'wilson-cowan',
#                                     "neuron_A": a,
#                                     "neuron_B": b,
#                                     "NT_A": nt1,
#                                     "NT_B": nt2,
#                                     "Z_A": G.nodes[a].get("zscore"),
#                                     "Z_B": G.nodes[b].get("zscore"),
#                                     "synapse": synapse,
#                                     "reciprocal_synapse": reciprocal
#                                 }

#     print(f"{GREEN}[INFO] Found {len(motifDict)} motifs with opposing NT polarity and reciprocal synapses{RESET}")
#     return motifDict


#     print(f"{GREEN}[INFO] Found {len(motifDict)} motifs with opposing NT polarity and reciprocal synapses{RESET}")
#     return motifDict
```

```python
import re
from collections import defaultdict

def find_edges_by_terminal_node(dot_path):
    edge_map = defaultdict(list)

    with open(dot_path, 'r') as f:
        for line in f:
            # Match basic edge syntax
            match = re.match(r'\s*"([^"]+)"\s*->\s*"([^"]+)"', line)
            if match:
                source, target = match.groups()
                # Extract final part if it's a compound edge
                simplified_source = source.split('--')[-1]
                simplified_target = target.split('--')[-1]
                key = (simplified_source, simplified_target)
                edge_map[key].append(line.strip())

    # Filter for multiple edges per neuron pair
    duplicates = {k: v for k, v in edge_map.items() if len(v) > 1}

    print(f"Found {len(duplicates)} terminal edge collisions:\n")
    for (src, tgt), lines in duplicates.items():
        print(f'Edges ending at "{tgt}" from "{src}": {len(lines)}')
        for l in lines:
            print(f'  {l}')
        print()

# Run it
find_edges_by_terminal_node('MBON13__35pre_001_002.gv')
```

```python
import re

def clean_graphviz_node_names(dot_str):
    # Match everything in double quotes
    all_strings = re.findall(r'"([^"]+)"', dot_str)

    # Remove strings clearly used as edge labels
    label_noise = re.compile(r"Zscore = \d+(\.\d+)?")
    filtered_nodes = [s for s in all_strings if not label_noise.match(s)]

    # Keep only nodes that appear either in edge definitions or node blocks
    edge_pattern = re.compile(r'"([^"]+)"\s*->\s*"([^"]+)"')
    node_names_in_edges = set(sum(edge_pattern.findall(dot_str), ()))

    # Node candidates that look like IDs and appear in edge mappings
    node_names = [n for n in filtered_nodes if n in node_names_in_edges]

    print(f"🧠 Total Nodes Rendered: {len(node_names)}")
    print(node_names[:50], '...')

    return node_names

def get_graphviz_edges(dot_str):
    edge_pattern = re.compile(r'"([^"]+)"\s*->\s*"([^"]+)"')
    edges = edge_pattern.findall(dot_str)

    print(f"\n🔗 Total Edges Rendered: {len(edges)}")
    for src, dst in edges[:10]:
        print(f"{src} → {dst}")
    
    return edges
```
