# Begin Synaptic data aggregation

```python


# ⏱️ Timer start
start_time = time.time()

neurons = my_client.NLP_result.neurons
hashids = list(neurons.keys())
names = [neurons[k]['uname'] for k in hashids]

# 🧠 Synapse collection
preSynapses = []
postSynapses = []

for i, hashid in enumerate(hashids):
    res = my_client.getInfo(hashid)
    data = res["data"]

    if "connectivity" in data:
        presyn = data["connectivity"]["pre"]["details"]
        postsyn = data["connectivity"]["post"]["details"]

        for syn in presyn:
            preSynapses.append(("pre", syn["syn_uname"], names[i], syn["uname"],
                                syn["s_rid"], syn["number"], True))

        for syn in postsyn:
            postSynapses.append(("post", syn["syn_uname"], names[i], syn["uname"],
                                 syn["s_rid"], syn["number"], True))

# 🧩 NumPy structured arrays
dtype = [('type', 'U4'), ('syn_uname', 'U100'), ('neuron_name', 'U100'),
         ('uname', 'U100'), ('s_rid', 'U100'), ('number', 'f8'), ('valid', 'bool')]

pre_array = np.array(preSynapses, dtype=dtype)
post_array = np.array(postSynapses, dtype=dtype)

# 🔍 Filter out 'segment' entries efficiently
pre_array = pre_array[np.char.find(pre_array['uname'], 'segment') == -1]
post_array = post_array[np.char.find(post_array['uname'], 'segment') == -1]

connectivity = {"hashids": hashids, "names": names, "pre_array": pre_array, "post_array": post_array}

# 📊 Dictionary construction
postSynNumberDict = {}
preSynNumberDict = {}

def process_array(array, synNumberDict, synType):
    neuron_names = np.unique(array['neuron_name'])
    for neuron in neuron_names:
        mask = array['neuron_name'] == neuron
        syn_data = array[mask]
        syn_numbers = syn_data['number']

        if syn_numbers.size == 0:
            continue

        z_scores = stats.zscore(syn_numbers)

        if np.any(np.isnan(z_scores)):
            array['valid'][mask] = False
            print(f'\033[93m{neuron} Error: Nan Value Found in array\033[0m')
            continue

        zmax = np.max(z_scores)
        norm_scores = 1 - ((zmax - z_scores) / zmax)

        print(f'\033[92m{neuron}: {synType}SynZmax= {zmax}\033[0m')

        for uname, rid, z, norm in zip(syn_data['syn_uname'], syn_data['s_rid'], z_scores, norm_scores):
            synNumberDict[rid] = [uname, z, norm]

process_array(connectivity["post_array"], postSynNumberDict, "post")
process_array(connectivity["pre_array"], preSynNumberDict, "pre")

# 🕒 Timing metrics
numberOfNodes = len(names)
end_time = time.time()
elapsed_time = end_time - start_time

avSynCalc_time = elapsed_time / (len(pre_array) + len(post_array))
avNeuronCalc_time = elapsed_time / numberOfNodes

avSynCalc_time = 10

if elapsed_time > avSynCalc_time:
    gvLayoutEngine = 'neato'
    gGraphvizMode = 'sgd'
else:
    gvLayoutEngine = 'sfdp'
    gGraphvizMode = 'spring'

print ("No. of HashIDs= " + str( len(hashids)))
print ("No. of names= " + str( numberOfNodes ))
print ("No. of preSynapses= " + str( len(preSynapses)))
print ("No. of postSynapses= " + str( len(postSynapses)))
print(f"Elapsed time: {elapsed_time:.2f} seconds")
print(f"Avg. time per synapse: {avSynCalc_time:.4f} seconds")
print(f"Avg. time per neuron: {avNeuronCalc_time:.4f} seconds")
print(f"Graphviz  set to Engine: {gvLayoutEngine} and Mode:{gGraphvizMode} ")
```

```python
def add_graph_edge(g, head, tail, zscore=None,
                    penwidth=None,
                    color=None,
                    edgeFontcolor="#40e0d0",
                    edgeFontName = "Ariel",
                    edgeFontSize = "5.0",
                    labelfloat = "false",
                    splines = "true",
                    arrowsize="0.4",
                    weight="3",
                    label_prefix="Zscore = ",
                    zScoreMultiplyer=2,
                    calcColorFn=None):
    """
    Adds a styled edge to the Graphviz graph object `g`.

    Parameters:
        g (Digraph): your graph object
        head (str): source node name
        tail (str): destination node name
        zscore (float): value to encode visually (used to compute penwidth and color)
        penwidth (float or str): overrides computed penwidth if provided
        color (str): overrides computed color if provided
        fontcolor (str): label font color
        fontname (str): label font name
        fontsize (str): label font size
        arrowsize (str): edge arrow size
        weight (str): edge weight
        label_prefix (str): prefix for edge label text
        labelfloat (str): whether label floats with edge curvature
        zScoreMultiplyer (float): multiplier for visual scaling
        calcColorFn (function): function to compute [penwidth, color] from zscore
    """

    # Auto-compute visual attributes if not provided
    if zscore is not None and calcColorFn:
        computed = calcColorFn(zscore)
        pw = penwidth if penwidth is not None else computed[0] * zScoreMultiplyer
        pc = color if color is not None else computed[1]
        label = label_prefix + str(round(computed[0], 2))
    else:
        pw = penwidth or "1.0"
        pc = color or "gray"
        label = ""
    #print(f"\033[92mcreating edge with head:\033[38;5;208m{head}\033[92m -> \033[96m{tail}\033[0m")
    g.edge(str(head), str(tail),
           labelfloat=labelfloat,
           label=label,
           fontsize=edgeFontSize,
           fontcolor=edgeFontcolor,
           fontname=edgeFontName,
           penwidth=str(pw),
           weight=weight,
           arrowsize=arrowsize,
           color=pc)
    
def add_edge_tograph(graph, head_name, tail_name, zscore, color_fn):
    """
    Wrapper for add_graph_edge to simplify synaptic connection additions.

    Parameters:
        graph: Graph object to modify.
        pre_name: Node name representing presynaptic neuron.
        post_name: Node name representing postsynaptic neuron.
        zscore: Z-score value representing synaptic weight.
        color_fn: Function to compute edge color from z-score.

    Returns:
        None
    """
    add_graph_edge(
        g=graph,
        head=head_name,
        tail=tail_name,
        zscore=zscore,
        calcColorFn=color_fn
    )
```

```python
# Add Edges to Graph

pw_default = 0.1
pc_default = 'gray'
minNormZscore = 0.03
zScoreMultiplyer = 2

skipped_edges_log = {}
nodeDegreeDict = {}
nodeNameZscoreDict = {}
zScoreMultiplyer = 2
valid_nodes = set()
unique_synapse = set()
unique_legs = set()
unused_legs = set()
orphaned_synapses = set()
skipped_edges_log = {}
unused_legs_log = {}

preSynapse = 'preSynapse'
postSynapse = 'postSynapse'

#Create data type for numpy searches
legDtype = [
    ('connection_type', 'U20'),
    ('head_name', 'U60'),
    ('tail_name', 'U60'),
    ('value', 'f8')
]


for head, tail, data in G.edges(data=True):
    
    
    #print('pre= ' + str(head) + 'post= ' + str(tail))
    head_name = G.nodes(data=True)[head]['uname']
    tail_name = G.nodes(data=True)[tail]['uname']
    valid_nodes.update([head_name])
    valid_nodes.update([tail_name])
    
    #if the head's name contains -- then this is a postsynapse syn->neuron
    if "--" in head_name:
        
        try:
            #IMPORTANT - The NEURONs presynaptic dict is used to get the synapses's postZ
            postZscore = preSynNumberDict[head][2]

            #use the head (postsynapse) as a synapse, the tail as a neuron
            #Note: If it's syn ->neuron, then this is a postsynapse
            unique_legs.add((postSynapse,head_name, tail_name, postZscore))
            unique_synapse.add((head,head_name))

        except (KeyError, TypeError, IndexError) as e:
            skipped_edges_log[f"⚠️MissingData {head}->{tail}"] = (
                f"{head} missing in preSynNumberDict or invalid structure: {repr(e)}"
            )
            continue  # skip and move to the next edge
        
        
        #print(f"Postsynaptic leg added with head hash {head}: {head_name} tail hash {tail}: {tail_name} and zscore= {postZscore} ")
        
    #if the tail's name contains -- then this is a presynapse neuron->syn        
    elif "--" in tail_name:
        
        try:      
        
            #IMPORTANT - The NEURONs postsynaptic dict is used to get the -synapses's- postZ
            preZscore = postSynNumberDict[tail][2]

            #use the head as a synapse, the tail here is a neuron node
            unique_legs.add((preSynapse,head_name, tail_name, preZscore))
            unique_synapse.add((tail,tail_name))
        
        except (KeyError, TypeError, IndexError) as e:
            skipped_edges_log[f"⚠️MissingData {head}->{tail}"] = (
                f"{tail} missing in postSynNumberDict or invalid structure: {repr(e)}"
            )
            continue  # skip and move to the next edge
        
        #print(f"Pretsynaptic leg added with head hash {head}: {head_name} tail hash {tail}: {tail_name} and zscore= {preZscore} ")

    else:
        skipped_edges_log[f"{head}->{tail}"] = f"{head_name} and {tail_name} not found"   

        
#Make an np array of all the synapse legs
synLegs_data = np.array(list(unique_legs), dtype=legDtype)

#Now loop through each unique synapse
for uniqueSynHash, uniqueSynName  in unique_synapse:   

        
    #Get the PRE synaptic LEG
    preSynResult = synLegs_data[
        (synLegs_data['tail_name'] == uniqueSynName ) & 
        (synLegs_data['connection_type'] == preSynapse )
    ]
    
    #if we find data, store the tuple into variables
    if len(preSynResult) == 1:
        preSynConnection, headNamePre, tailNamePre, preZscore = preSynResult[0]
    else:
        skipped_edges_log[f"⚠️Error{uniqueSynHash}-{uniqueSynName}"] = f"Edge for {uniqueSynName} not found" 
        
    
    #Get the POST synaptic LEG
    postSynResult = synLegs_data[
        (synLegs_data['head_name'] == uniqueSynName ) & 
        (synLegs_data['connection_type'] == postSynapse )
    ]

    if len(postSynResult) == 1:
        postSynConnection, headNamePost , tailNamePost, postZscore = postSynResult[0]
    else:
        skipped_edges_log[f"⚠️Error{uniqueSynHash}-{uniqueSynName}"] = f"Edge for {uniqueSynName} not found" 
    
    #we should have two legs here
    #pdb.set_trace()
    
    
    #We only want to process associated pre and post legs if they fall below the min
    if  preZscore >= minNormZscore and postZscore >= minNormZscore: 
    
        #Add pre and post legs 
        add_edge_tograph(g, headNamePre, tailNamePre, preZscore, calcColor)
        add_edge_tograph(g, headNamePost, tailNamePost, postZscore, calcColor)

    else:
        
        
        #save the unused legs for.. reasons
        unused_legs.add((preSynConnection,headNamePre, tailNamePre,preZscore))
        unused_legs.add((postSynConnection,headNamePost, tailNamePost,postZscore))
        unused_legs_log[f"{headNamePre}->{tailNamePre}"] = f"Ignore leg with preZ={preZscore} and postZ={postZscore}"
        
        # Use the pre tailname to note which syanaptic nodes can be removed
        orphaned_synapses.add(tailNamePre)
        
        
print(f"⚠️ Error with {len(skipped_edges_log)} edges not found! ")
display(JSON(skipped_edges_log, expanded=False))
print(f"⚠️ [INFO] {len(unused_legs_log)} edges unused due to low Z-scores")
display(JSON(unused_legs_log, expanded=False))

print( f"number of legs: {len(unique_legs)}" )
# print(f" Valid legs: {unique_legs}")


print(f"number of G nodes: {len(G.nodes())}")
print(f"number of valid_nodes nodes: {len(valid_nodes)}")
```

```python
import warnings

def clean_node_id(node_str):
    return node_str.strip('"').strip()

def parse_edge_line(line):
    pattern = r'\"?([^\s\"]+)\"?\s*->\s*\"?([^\s\"]+)\"?.*?label=\"Zscore\s*=\s*([-+]?[0-9]*\.?[0-9]+)\"'
    match = re.search(pattern, line.strip())
    if match:
        head, tail, z_score = match.groups()
        return head.strip('"'), tail.strip('"'), float(z_score), line.strip()
    return None


def merge_graph_files(directory='merge_graphs', output_file='merged_data.gv'):
    if not os.path.exists(directory):
        os.makedirs(directory)
        warnings.warn("Directory 'merge_graphs' not found. Created empty directory. Halting routine.")
        return

    merged_edges = []
    graph_header = []

    for filename in os.listdir(directory):
        if filename.endswith('.gv') and not filename.startswith('merged') and not filename.startswith('unused'):
            with open(os.path.join(directory, filename), 'r') as f:
                lines = f.readlines()

            if not graph_header:
                graph_header = lines[:2]

            body_lines = []  # 🔄 Reset for each file
            layout_directives = ('size=', 'rotation=', 'splines=')

            for line in lines[2:]:
                stripped = line.strip()
                if stripped == '}':
                    continue
                # Only skip lines that *start* with layout directives
                if any(stripped.startswith(directive) for directive in layout_directives):
                    print(f"Skipping layout directive line: {stripped}")
                    continue
                body_lines.append(line)

                
            merged_edges.extend(body_lines)
            print(f"{len(body_lines)} lines of file {filename} merged. Total lines in merged now equals: {len(merged_edges)}")

    # Deduplicate after all files are processed
    edge_dict = {}
    unused_edges = {}
    node_buffer = []
    node_set = set()

    for line in merged_edges:
        parsed = parse_edge_line(line)

        if parsed:
            #print(f"Parsed result: --{parsed}--, Line = {line}")
            parsed_head, parsed_tail, z, full_line = parsed
            key = (parsed_head, parsed_tail)

            # Filtering logic: keep only highest Z-score
            if key not in edge_dict or z > edge_dict[key][0]:
                edge_dict[key] = (z, full_line)
            else:
                # Log discarded edge
                unused_edges.setdefault(key, []).append((z, full_line))
        else:
            # Node handling logic (as discussed earlier)
            line = line.strip()
            if not line:
                continue
            node_buffer.append(line)
            if line.endswith(']'):
                full_node = ' '.join(node_buffer)
                node_set.add(full_node)
                node_buffer = []
                #print(f"Parsed Node found: {full_node}")



                
    output_path_merged = os.path.join(directory, output_file)

    #pdb.set_trace()
    #Write out data to merged file
    with open(output_path_merged, 'w') as out:
        out.writelines(graph_header)
        # Write unique nodes
        for node_line in sorted(node_set):
            out.write(node_line + '\n')
        # Write edges
        for _, (_, line) in edge_dict.items():
            out.write(line + '\n')
        out.write('\tsize="50,50"\n')
        out.write('}\n')

    print(f"✅ Merged graph written to {output_file} with {len(edge_dict)} unique edges.")

    output_path_log = os.path.join(directory, 'unused_log.gv')
    
    with open(output_path_log, 'w') as f:
        for key, entries in unused_edges.items():
            for z_val, line_text in entries:
                f.write(f"# Discarded [{key[0]} -> {key[1]}] with z={z_val}: {line_text}\n")
    print(f"✅ Unused itmes log written to {output_path_log} with {len(unused_edges)}  items.")

############RENAME IF Needed
# currentMergedFileName = 'merged_data.gv'
currentMergedFileName = 'MBON03_prePost_merged_data.gv'

merge_graph_files(output_file=currentMergedFileName)
```

## Python routine that scans the G_auto.gv file-identifies all edges with the same head and tail. This can pinpoint which ones might be collapsed due to concentrate=true.

```python
import re
from collections import defaultdict

def find_edges_by_terminal_node(dot_path):
    edge_map = defaultdict(list)

    with open(dot_path, 'r') as f:
        for line in f:
            # Match basic edge syntax
            match = re.match(r'\s*"([^"]+)"\s*->\s*"([^"]+)"', line)
            if match:
                source, target = match.groups()
                # Extract final part if it's a compound edge
                simplified_source = source.split('--')[-1]
                simplified_target = target.split('--')[-1]
                key = (simplified_source, simplified_target)
                edge_map[key].append(line.strip())

    # Filter for multiple edges per neuron pair
    duplicates = {k: v for k, v in edge_map.items() if len(v) > 1}

    print(f"Found {len(duplicates)} terminal edge collisions:\n")
    for (src, tgt), lines in duplicates.items():
        print(f'Edges ending at "{tgt}" from "{src}": {len(lines)}')
        for l in lines:
            print(f'  {l}')
        print()

# Run it
find_edges_by_terminal_node('MBON13__35pre_001_002.gv')
```

```python
import re

def clean_graphviz_node_names(dot_str):
    # Match everything in double quotes
    all_strings = re.findall(r'"([^"]+)"', dot_str)

    # Remove strings clearly used as edge labels
    label_noise = re.compile(r"Zscore = \d+(\.\d+)?")
    filtered_nodes = [s for s in all_strings if not label_noise.match(s)]

    # Keep only nodes that appear either in edge definitions or node blocks
    edge_pattern = re.compile(r'"([^"]+)"\s*->\s*"([^"]+)"')
    node_names_in_edges = set(sum(edge_pattern.findall(dot_str), ()))

    # Node candidates that look like IDs and appear in edge mappings
    node_names = [n for n in filtered_nodes if n in node_names_in_edges]

    print(f"🧠 Total Nodes Rendered: {len(node_names)}")
    print(node_names[:50], '...')

    return node_names

def get_graphviz_edges(dot_str):
    edge_pattern = re.compile(r'"([^"]+)"\s*->\s*"([^"]+)"')
    edges = edge_pattern.findall(dot_str)

    print(f"\n🔗 Total Edges Rendered: {len(edges)}")
    for src, dst in edges[:10]:
        print(f"{src} → {dst}")
    
    return edges
```
